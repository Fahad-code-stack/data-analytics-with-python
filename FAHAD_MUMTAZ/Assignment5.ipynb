#Question 1

```python
numbers = [12,4,45,76,45,50,5,111]

total_sum = 0
even_numbers = []

for num in numbers:
  print(num)

  total_sum += num

  if num % 2 == 0:
    even_numbers.append(num)

print("\n Sum of all numbers")
print(f"total_sum is: {total_sum}")
print("Even numbers")
for even in even_numbers:
  print(even)
```

 #Question 2

```python
Students = {
    "name": "Alex Mercer",
    "age": 20,
    "course": "Computer Science",
    "marks": [85, 45, 92]
}
print("Students Detyails")

for key, value in Students.items():
  print(f"{key}: {value}")
```

```python
total_marks = 0
number_of_marks = 0
for mark in Students["marks"]:
  total_marks += mark
  number_of_marks += 1

average_marks = total_marks / number_of_marks

print("Final Result")
print(f"Average Marks: {average_marks}")

if average_marks >= 50:
    print("Status: Passed!")
else:
    print("Status: Failed.")
```

#Question 3

```python
employees = [
    {"name": "Sarah Jenkins", "department": "HR", "salary": 55000},
    {"name": "David Chen", "department": "Engineering", "salary": 92000},
    {"name": "Elena Rodriguez", "department": "Marketing", "salary": 68000},
    {"name": "Marcus Johnson", "department": "Sales", "salary": 74000}
]

print("Employee Details")
for employee in employees:
  for key, value in employee.items():
    print(f"{key}: {value}")
  print()
total_salary_expense = 0
highest_salary = 0
highest_salary_employee = ""

for employee in employees:
  total_salary_expense += employee["salary"]

  if employee["salary"] > highest_salary:
    highest_salary = employee["salary"]
    highest_salary_employee = employee["name"]

print(f"Total Salary Expense: ${total_salary_expense:,}")
print(f"Highest Earner: {highest_salary_employee} with a salary of ${highest_salary:,}")
```

#Question 4

```python
numbers = []
print("Enter number in list")

while True:
  num = int(input("Enter as number:"))

  if  num == -1:
      break

numbers.append(num)

print("Final Result")
print(f"List: {numbers}")

if numbers:
  print(f"Largest Numbers: {max(numbers)}")
  print(f"Smallest Number: {min(numbers)}")
else:
  print("No number was enter before stopping ")
```

#Question 5

```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()

word_counts = {}

for words in words:
  if words in word_counts:
    word_counts[words] += 1
  else:
    word_counts[words] = 1
print("Word Counts:")
for word, count in word_counts.items():
  print(f"{word}: {count}")
```
